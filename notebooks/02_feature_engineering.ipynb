# 02 Feature Engineering

This notebook loads the cleaned processed datasets and creates analytical features for exploratory analysis and Power BI dashboard development. It adds average salary, salary level, career stage, remote-friendly status, and role group columns, then saves the feature-engineered dataset back to the processed data folder.


```python
from pathlib import Path
import pandas as pd

def find_project_root(start_path):
    for path in [start_path, *start_path.parents]:
        if (path / "data" / "processed").exists():
            return path
    raise FileNotFoundError("Could not find project root with data/processed folder.")

BASE_DIR = find_project_root(Path.cwd())
PROCESSED_DATA_DIR = BASE_DIR / "data" / "processed"

ai_jobs = pd.read_csv(PROCESSED_DATA_DIR / "ai_jobs_cleaned.csv")
skills_demand = pd.read_csv(PROCESSED_DATA_DIR / "skills_demand_cleaned.csv")
country_trends = pd.read_csv(PROCESSED_DATA_DIR / "country_trends_cleaned.csv")

print("Cleaned datasets loaded successfully")
print("ai_jobs:", ai_jobs.shape)
print("skills_demand:", skills_demand.shape)
print("country_trends:", country_trends.shape)
```

```python
print("ai_jobs data types:")
print(ai_jobs.dtypes)

print("\nskills_demand data types:")
print(skills_demand.dtypes)

print("\ncountry_trends data types:")
print(country_trends.dtypes)
```

```python
ai_jobs["salary_avg_usd"] = (ai_jobs["salary_min_usd"] + ai_jobs["salary_max_usd"]) / 2

def categorize_salary(salary):
    if salary < 80000:
        return "Low"
    elif salary < 120000:
        return "Medium"
    elif salary < 150000:
        return "High"
    else:
        return "Very High"

ai_jobs["salary_level"] = ai_jobs["salary_avg_usd"].apply(categorize_salary)

ai_jobs[["salary_min_usd", "salary_max_usd", "salary_avg_usd", "salary_level"]].head()
```

```python
def create_career_stage(level):
    if level == "Entry":
        return "Entry Level"
    elif level == "Mid":
        return "Mid Level"
    elif level == "Senior":
        return "Senior Level"
    else:
        return "Unknown"

ai_jobs["career_stage"] = ai_jobs["experience_level"].apply(create_career_stage)

ai_jobs[["experience_level", "career_stage"]].head()
```

```python
def remote_friendly(remote_type):
    if remote_type in ["Remote", "Hybrid"]:
        return "Yes"
    elif remote_type == "Onsite":
        return "No"
    else:
        return "Unknown"

ai_jobs["is_remote_friendly"] = ai_jobs["remote_type"].apply(remote_friendly)

ai_jobs[["remote_type", "is_remote_friendly"]].head()
```

```python
def create_role_group(job_title):
    if job_title == "Data Analyst":
        return "Data Analytics"
    elif job_title == "Data Scientist":
        return "Data Science"
    elif job_title == "Machine Learning Engineer":
        return "Machine Learning"
    elif job_title == "MLOps Engineer":
        return "MLOps"
    elif job_title == "AI Researcher":
        return "AI Research"
    elif job_title == "Applied Scientist":
        return "Applied AI / Research"
    else:
        return "Other"

ai_jobs["role_group"] = ai_jobs["job_title"].apply(create_role_group)

ai_jobs[["job_title", "role_group"]].head()
```

```python
feature_columns = [
    "salary_avg_usd",
    "salary_level",
    "career_stage",
    "is_remote_friendly",
    "role_group"
]

print("Feature columns added:")
print(ai_jobs[feature_columns].head())

print("\nMissing values in feature columns:")
print(ai_jobs[feature_columns].isnull().sum())
```

```python
ai_jobs.to_csv(PROCESSED_DATA_DIR / "ai_jobs_cleaned.csv", index=False)
skills_demand.to_csv(PROCESSED_DATA_DIR / "skills_demand_cleaned.csv", index=False)
country_trends.to_csv(PROCESSED_DATA_DIR / "country_trends_cleaned.csv", index=False)

print("Feature-engineered datasets saved successfully")
print("ai_jobs:", ai_jobs.shape)
print("skills_demand:", skills_demand.shape)
print("country_trends:", country_trends.shape)
```

## Feature Engineering Summary

The cleaned job dataset now includes salary, career stage, remote-friendly, and role-group features. These processed files are ready for exploratory analysis, dashboard tables, and Power BI export.
