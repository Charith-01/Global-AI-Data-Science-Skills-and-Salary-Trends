# 04 Power BI Data Export

This notebook prepares the final datasets required for Power BI dashboard development. The feature-engineered and analysed data is reviewed, selected, and exported into clean CSV files that can be directly loaded into Power BI.

The exported files include the main job market dataset, the skills demand dataset, and the country trends dataset. These files are saved in the Power BI data folder and used to build the final interactive dashboard pages.

```python
from pathlib import Path
import pandas as pd

def find_project_root(start_path):
    for path in [start_path, *start_path.parents]:
        if (path / "data" / "processed").exists() and (path / "powerbi").exists():
            return path
    raise FileNotFoundError("Could not find project root with data/processed and powerbi folders.")

BASE_DIR = find_project_root(Path.cwd())

PROCESSED_DATA_DIR = BASE_DIR / "data" / "processed"
POWERBI_DATA_DIR = BASE_DIR / "powerbi" / "data"

POWERBI_DATA_DIR.mkdir(parents=True, exist_ok=True)

ai_jobs = pd.read_csv(PROCESSED_DATA_DIR / "ai_jobs_cleaned.csv")
skills_demand = pd.read_csv(PROCESSED_DATA_DIR / "skills_demand_cleaned.csv")
country_trends = pd.read_csv(PROCESSED_DATA_DIR / "country_trends_cleaned.csv")

print("Data loaded successfully")
print("Project root:", BASE_DIR)
print("Power BI data folder:", POWERBI_DATA_DIR)
```

```python
print("Processed data folder:")
for file in sorted(PROCESSED_DATA_DIR.iterdir()):
    print(file.name)
```

```python
print("Dataset shapes:")
print("ai_jobs:", ai_jobs.shape)
print("skills_demand:", skills_demand.shape)
print("country_trends:", country_trends.shape)
```

```python
ai_jobs.to_csv(POWERBI_DATA_DIR / "ai_jobs_powerbi.csv", index=False)
skills_demand.to_csv(POWERBI_DATA_DIR / "skills_demand_powerbi.csv", index=False)
country_trends.to_csv(POWERBI_DATA_DIR / "country_trends_powerbi.csv", index=False)

print("Power BI CSV files exported successfully")
```

```python
print("Power BI data files:")

for file in sorted(POWERBI_DATA_DIR.iterdir()):
    print(file.name)
```

## Power BI Data Preparation Summary

After completing data cleaning, feature creation, exploratory analysis, and Matplotlib visualizations, the final processed datasets were exported for Power BI dashboard development.

Three Power BI-ready CSV files were created:

- `ai_jobs_powerbi.csv`
- `skills_demand_powerbi.csv`
- `country_trends_powerbi.csv`

The `ai_jobs_powerbi.csv` file contains cleaned job posting data with additional analytical features such as average salary, salary level, career stage, role group, and remote-friendly classification.

These exported files will be used to build an interactive Power BI dashboard with filters, visuals, and business insights.
